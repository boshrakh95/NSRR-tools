# xfig_14 — Task Similarity Clustermap

Clusters tasks by the **shape of their saturation curve** (AUROC at each
context length as a 7-D feature vector).

Rows = tasks (hierarchically clustered); columns = context lengths (fixed order).
Color = AUROC. Tasks with similar curves are placed near each other.

Idea #14 from `docs/NEW_PLOT_IDEAS.md`.

**Data**: `analysis.csv`, k=all.
